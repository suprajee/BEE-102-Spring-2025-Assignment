**Building Markov transition matrix** 

```python
import pandas as pd

# Sample input sequence
sequence = "ACGTACAGATACAGATACGCTCTCGACTG"
alphabet = ['A', 'C', 'G', 'T']

transition_counts = {s: {t: 0 for t in alphabet} for s in alphabet}

# Count dinucleotides: For each position, count transition from current base to next base
for i in range(len(sequence) - 1):
    current = sequence[i]
    next_state = sequence[i+1]
    if current in alphabet and next_state in alphabet:
        transition_counts[current][next_state] += 1

# Build the Markov transition matrix by normalizing counts for each state (row)
transition_matrix = {}
for state in alphabet:
    row_sum = sum(transition_counts[state].values())
    if row_sum > 0:
        transition_matrix[state] = {t: transition_counts[state][t] / row_sum for t in alphabet}
    else:
        transition_matrix[state] = {t: 0 for t in alphabet}

df = pd.DataFrame(transition_matrix).T
print("Markov Transition Matrix:")
print(df)
```
