## Multi-line sequence fasta to single-line fasta file
1. **File Handling**: 
  - Opens the input file "multiline_input.fasta" for reading
  - Opens a new output file "single_line_output.fasta" for writing
  - Uses a with-statement to ensure proper file closure

2. **Tracking First Record**:
  - Uses a `first_record` flag to handle special formatting for the first sequence

3. **Line-by-Line Processing**:
  - For each line in the input file:
    - Removes trailing newline characters with `rstrip("\n")`
    - Skips empty lines with the `continue` statement

4. **Header Line Handling**:
  - Identifies header lines that start with ">"
  - For all but the first record, adds a newline character before writing the header
    (This ensures each sequence ends properly before starting a new header)
  - Writes the header followed by a newline

5. **Sequence Line Handling**:
  - For sequence lines (any line not starting with ">")
  - Writes the sequence content directly without adding newlines
  - This effectively concatenates multiple sequence lines into a single continuous line


<br>Single-line fasta is saved as single_line_output.fasta


```python
with open("multiline_input.fasta", "r") as infile, open("single_line_output.fasta", "w") as outfile:
    first_record = True
    for line in infile:
        # Remove any trailing newline characters
        line = line.rstrip("\n")
        if not line:
            continue 
        if line.startswith(">"):
            # If we are not at the first record, then finish the previous sequence by writing a newline.
            if not first_record:
                outfile.write("\n")
            else:
                first_record = False
            outfile.write(line + "\n")
        else:
            # For sequence lines, write them directly (without adding a newline)
            outfile.write(line)
    
print("Single-line FASTA file created: single_line_output.fasta")
```
