**Multi-line sequence fasta to single-line fasta file**
<br>Single-line fasta is saved as single_line_output.fasta

```python
with open("multiline_input.fasta", "r") as infile, open("single_line_output.fasta", "w") as outfile:
    first_record = True
    for line in infile:
        # Remove any trailing newline characters
        line = line.rstrip("\n")
        if not line:
            continue 
        if line.startswith(">"):
            # If we are not at the first record, then finish the previous sequence by writing a newline.
            if not first_record:
                outfile.write("\n")
            else:
                first_record = False
            outfile.write(line + "\n")
        else:
            # For sequence lines, write them directly (without adding a newline)
            outfile.write(line)
    
print("Single-line FASTA file created: single_line_output.fasta")
```
